## 1. 集中趋势度量：均值、中位数、众数

**问题**：这组数据的中心在哪里？

```python
scores = np.array([78, 85, 92, 68, 88])           # 5 名学生的期末成绩
n = len(scores)                                   # 样本量 n = 5

mean_manual = np.sum(scores) / n                  # 手动计算均值：总和 ÷ 个数
median_manual = np.median(scores)                 # 中位数：排序后取正中间的值

print(f'成绩: {scores}')                          # 打印原始数据
print(f'均值 = {mean_manual}')                     # 打印均值（应为 82.2）
print(f'中位数 = {median_manual}')                 # 打印中位数
print(f'排序后: {np.sort(scores)}')                # 排序验证中位数位置
```

```python
# ===== 均值的致命弱点：极端值敏感 =====
scores_outlier = np.array([78, 85, 92, 68, 200])  # 把 88 改成 200（极端值）
mean_outlier = np.sum(scores_outlier) / 5         # 重新计算均值

print(f'含极端值的均值 = {mean_outlier}')           # 均值被拉高到 104.6
print(f'含极端值的中位数 = {np.median(scores_outlier)}')   # 中位数几乎不变——抗干扰
print('结论：有极端值时用中位数，对称分布用均值')     # 选择建议
```

```python
# ===== 众数：出现次数最多的值（也适用于分类数据）=====
brands = pd.Series(['苹果', '华为', '苹果', '小米', '苹果', '华为'])   # 手机品牌偏好（名义数据）
print('品牌众数:', brands.mode()[0])               # mode() 返回众数（可能多个，取第一个）
print(brands.value_counts())                      # value_counts() 统计各品牌频数
```

```python
# ===== 加权均值：不同数据赋予不同权重 =====
weights = np.array([0.20, 0.30, 0.50])            # 权重：作业 20%、期中 30%、期末 50%
scores3 = np.array([85, 78, 92])                  # 三项成绩

weighted_mean = np.sum(weights * scores3) / np.sum(weights)   # 加权均值公式：Σ(w·x) / Σw
print(f'加权平均成绩 = {weighted_mean}')            # 应为 86.4（不是简单平均 85）
print(f'对比简单平均 = {np.mean(scores3)}')         # 简单平均忽略了权重差异
```

## 2. 离散程度度量：从方差到变异系数

**核心思想**：均值相同的数据，离散程度可能完全不同。

```python
data = np.array([4, 8, 6, 10, 12])                # 例题数据（n = 5）
n = len(data)                                     # 样本量
mean_val = np.sum(data) / n                       # Step 1：均值 = 40/5 = 8

deviations = data - mean_val                      # Step 2：每个点与均值的偏差 d = x - x̄
squared_devs = deviations ** 2                    # Step 3：偏差平方（消除正负抵消）
SS = np.sum(squared_devs)                         # Step 4：平方偏差和 SS = 40
var_sample = SS / (n - 1)                         # Step 5：样本方差 = SS/(n-1) = 40/4 = 10

print(f'均值 = {mean_val}')                        # 打印均值
print(f'偏差 = {deviations}（和为 0）')             # 验证偏差之和恒为 0
print(f'平方偏差 = {squared_devs}')                 # 打印各点平方偏差
print(f'样本方差 s² = {var_sample}')                # 打印样本方差（分母 n-1）
print(f'样本标准差 s = {np.sqrt(var_sample):.4f}')  # 开方得标准差 ≈ 3.16（与原数据同单位）
```

```python
# ===== 为什么除以 n-1？对比两种分母 =====
var_n = SS / n                                    # 除以 n：会低估总体方差
var_n1 = SS / (n - 1)                             # 除以 n-1：无偏估计

print(f'除以 n   的方差 = {var_n}')                 # 有偏（偏小）
print(f'除以 n-1 的方差 = {var_n1}')                # 无偏
print('原因：样本均值天然贴合样本点，偏差和偏小；')   # 直观解释 1
print('用样本均值代替总体均值损失了 1 个自由度')      # 直观解释 2
```

```python
# ===== 用内置函数验证（注意 ddof 参数）=====
print(f'np.var 默认(ddof=0): {np.var(data):.4f}')          # 默认除以 n —— 总体方差
print(f'np.var ddof=1      : {np.var(data, ddof=1):.4f}')  # ddof=1 除以 n-1 —— 样本方差
print(f'np.std ddof=1      : {np.std(data, ddof=1):.4f}')  # 样本标准差（同单位）
```

### 经验法则（近似正态分布）
- 约 **68%** 数据落在 $(\bar{x}-s,\ \bar{x}+s)$ 内
- 约 **95%** 数据落在 $(\bar{x}-2s,\ \bar{x}+2s)$ 内
- 约 **99.7%** 数据落在 $(\bar{x}-3s,\ \bar{x}+3s)$ 内

```python
# ===== 变异系数 CV：消除量级影响，才能公平比较 =====
company_A = {'mean': 10000, 'std': 2000}          # A 公司：月薪均值 1 万、标准差 2000
company_B = {'mean': 50000, 'std': 5000}          # B 公司：月薪均值 5 万、标准差 5000

cv_A = company_A['std'] / company_A['mean'] * 100 # CV = s/x̄ × 100%（A 公司）
cv_B = company_B['std'] / company_B['mean'] * 100 # CV = s/x̄ × 100%（B 公司）

print(f'A 公司 CV = {cv_A:.1f}%')                   # 20%
print(f'B 公司 CV = {cv_B:.1f}%')                   # 10%
print('结论：A 公司工资相对差异更大（绝对标准差小≠差异小）')   # CV 的价值
print('注意：CV 只能用于比率数据（有绝对零点的正值）')        # 使用限制
```

## 3. Pandas 一键描述性统计

```python
np.random.seed(42)                                # 固定随机种子
n_students = 50                                   # 模拟 50 名学生

df = pd.DataFrame({                               # 创建成绩表
    '数学': np.random.normal(75, 12, n_students).round(1),    # 数学：均值 75、标准差 12
    '英语': np.random.normal(78, 10, n_students).round(1),    # 英语：均值 78、标准差 10
    '物理': np.random.normal(72, 15, n_students).round(1),    # 物理：均值 72、标准差 15
})

df.describe()                                     # 一键输出 count/mean/std/min/四分位数/max
```

## 4. 可视化：直方图与箱线图

```python
subjects = ['数学', '英语', '物理']                 # 三个科目
colors = ['#C49A6C', '#8B6F4E', '#E8D5C4']        # 三种颜色

fig, axes = plt.subplots(1, 3, figsize=(15, 4))   # 1 行 3 列的子图布局
for i, (subj, color) in enumerate(zip(subjects, colors)):   # 遍历科目与颜色
    axes[i].hist(df[subj], bins=12, color=color, edgecolor='white')   # 直方图分 12 个区间
    axes[i].axvline(df[subj].mean(), color='#3D3D3D', linestyle='--')   # 均值处画虚线
    axes[i].set_title(f'{subj}成绩分布')            # 子图标题
    axes[i].set_xlabel('分数')                      # x 轴标签
    axes[i].set_ylabel('人数')                      # y 轴标签
plt.tight_layout()                                # 自动调整间距防止重叠
plt.show()                                        # 显示图形
```

```python
# ===== 箱线图：一行代码看懂中位数、四分位距与异常值 =====
plt.figure(figsize=(10, 6))                        # 新建画布
plt.boxplot([df['数学'], df['英语'], df['物理']],  # 传入三科成绩数据
            tick_labels=['数学', '英语', '物理'],    # x 轴刻度标签（新版 matplotlib 用 tick_labels）
            patch_artist=True,                     # 允许给箱体填色
            boxprops=dict(facecolor='#E8D5C4'))    # 箱体颜色
plt.title('三科成绩箱线图对比')                     # 图表标题
plt.ylabel('分数')                                 # y 轴标签
plt.grid(True, alpha=0.3)                          # 显示 30% 透明度的网格
plt.show()                                        # 显示图形
# 读图：箱中线=中位数；箱边=Q1/Q3；箱高=IQR；须外散点=异常值
```

## 5. 实战：30 天销售数据分析

```python
np.random.seed(123)                               # 固定种子
days = 30                                         # 30 天数据
sales = np.random.normal(50, 15, days)            # 模拟日销售额：均值 50 万、标准差 15 万
sales = np.maximum(sales, 5)                      # 保证销售额不低于 5 万（截断负值）
sales[15] = 120                                   # 人为设置第 16 天为促销日极端值 120 万

mean_s, median_s = np.mean(sales), np.median(sales)   # 均值与中位数
std_s = np.std(sales, ddof=1)                     # 样本标准差
cv_s = std_s / mean_s * 100                       # 变异系数

print(f'均值 = {mean_s:.2f} 万元')                  # 打印均值
print(f'中位数 = {median_s:.2f} 万元')              # 打印中位数
print(f'标准差 = {std_s:.2f} 万元')                 # 打印标准差
print(f'变异系数 = {cv_s:.1f}%')                    # 打印 CV
print(f'均值 > 中位数 → 右偏（促销极端值拉高均值）')   # 偏态判断：右偏
```

## 6. 本周小结

| 度量 | 公式 | 适用场景 |
|---|---|---|
| 均值 | $\bar{x}=\frac{\sum x_i}{n}$ | 对称分布、无异常值 |
| 中位数 | 排序后中间值 | 偏态分布、有异常值 |
| 众数 | 出现最多 | 分类数据 |
| 样本方差 | $s^2=\frac{\sum(x_i-\bar{x})^2}{n-1}$ | 无偏估计（注意 ddof=1）|
| 变异系数 | $CV=\frac{s}{\bar{x}}\times 100\%$ | 跨量级比较（仅比率数据）|
