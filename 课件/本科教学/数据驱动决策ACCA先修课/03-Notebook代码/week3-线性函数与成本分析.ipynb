## 1. 线性成本函数：$y = a + bx$

- 截距 $a$ = **固定成本**（不生产也要支出，如租金）
- 斜率 $b$ = **单位变动成本**（每多生产一件增加的成本）

```python
a, b = 10000, 50                                  # 固定成本 1 万元/月，单位变动成本 50 元/件
x = 200                                           # 本月产量 200 件
y = a + b * x                                     # 总成本 = 10000 + 50×200
print(f'产量 {x} 件时的总成本 = {y} 元')           # 应为 20000 元
```

```python
# ===== 两点式求成本方程：b = Δy/Δx，a = y₁ - b·x₁ =====
x1, y1 = 100, 15000                               # 1月：产量 100 件，总成本 15000 元
x2, y2 = 300, 25000                               # 2月：产量 300 件，总成本 25000 元

b_hat = (y2 - y1) / (x2 - x1)                     # 斜率 = (25000-15000)/(300-100) = 50
a_hat = y1 - b_hat * x1                           # 截距 = 15000 - 50×100 = 10000

print(f'斜率 b（单位变动成本）= {b_hat} 元/件')     # 打印斜率
print(f'截距 a（固定成本）= {a_hat} 元')            # 打印截距
print(f'成本方程：y = {a_hat} + {b_hat}x')          # 打印完整方程
```

```python
# ===== 不同成本结构的对比：高固定低变动 vs 低固定高变动 =====
x = np.linspace(0, 400, 100)                      # 产量范围 0~400，生成 100 个点
y1 = 10000 + 50 * x                               # 方案1：固定 1 万、变动 50/件
y2 = 8000 + 60 * x                                # 方案2：固定 8 千、变动 60/件（低固定高变动）
y3 = 12000 + 40 * x                               # 方案3：固定 1.2 万、变动 40/件（高固定低变动）

plt.figure(figsize=(10, 6))                        # 新建画布
plt.plot(x, y1, label='a=10000, b=50', color='#C49A6C', linewidth=2)   # 画方案1成本线
plt.plot(x, y2, label='a=8000, b=60', color='#8B6F4E', linewidth=2)    # 画方案2成本线
plt.plot(x, y3, label='a=12000, b=40', color='#E8A87C', linewidth=2)   # 画方案3成本线
plt.xlabel('产量 x（件）')                          # x 轴标签
plt.ylabel('总成本 y（元）')                        # y 轴标签
plt.title('不同成本结构下的总成本曲线')              # 标题
plt.legend()                                      # 显示图例
plt.grid(True, alpha=0.3)                         # 网格
plt.show()                                        # 显示
# 观察交点：低产量时方案2便宜；高产量时方案3更划算——成本结构决定竞争策略
```

## 2. 高低点法：分解半变动成本

半变动成本（如电费）= 固定部分 + 变动部分。高低点法用**活动量最高和最低**两个点求解。

```python
months = ['1月', '2月', '3月', '4月', '5月', '6月']                 # 月份标签
x = np.array([200, 250, 180, 320, 280, 150])      # 各月产量（件）
y = np.array([5800, 6500, 5500, 7600, 6900, 5100])   # 各月电费（元）

h_idx = np.argmax(x)                              # argmax 找到产量最高月的下标（4月）
l_idx = np.argmin(x)                              # argmin 找到产量最低月的下标（6月）
x_h, y_h = x[h_idx], y[h_idx]                     # 最高点 (320, 7600)
x_l, y_l = x[l_idx], y[l_idx]                     # 最低点 (150, 5100)

b_elec = (y_h - y_l) / (x_h - x_l)                # 变动成本率 = Δy/Δx = 2500/170 ≈ 14.71
a_elec = y_h - b_elec * x_h                       # 固定成本 = 7600 - 14.71×320 ≈ 2893

print(f'最高点: ({x_h}, {y_h})（{months[h_idx]}）')  # 打印最高点
print(f'最低点: ({x_l}, {y_l})（{months[l_idx]}）')  # 打印最低点
print(f'变动成本率 b = {b_elec:.2f} 元/件')          # 打印斜率
print(f'固定成本 a = {a_elec:.2f} 元')               # 打印截距
print(f'电费方程: y = {a_elec:.2f} + {b_elec:.2f}x') # 打印成本方程
```

```python
# ===== 用拟合出的方程预测 =====
x_new = 300                                       # 预测产量 300 件时的电费
y_pred = a_elec + b_elec * x_new                  # 代入成本方程
print(f'产量 300 件的预测电费 = {y_pred:.0f} 元')   # 打印预测值

print('优缺点提醒：')                               # 高低点法评价
print('✓ 计算简单，只需两个数据点')                  # 优点
print('✗ 只用两个极端点，忽略其他数据，易受异常值影响') # 缺点
```

## 3. 散点图与最佳拟合线

```python
# ===== 产量 vs 总成本散点图 + 拟合线 =====
x2 = np.array([100, 150, 200, 250, 300, 350, 400])           # 产量数据
y2 = np.array([12000, 14000, 16500, 19000, 21000, 24000, 26000])   # 总成本数据

coef = np.polyfit(x2, y2, 1)                      # 一次多项式拟合，返回 [斜率b, 截距a]
b_fit, a_fit = coef[0], coef[1]                   # 拆分斜率和截距

x_line = np.linspace(x2.min(), x2.max(), 100)     # 生成画线的 x 值序列
y_line = a_fit + b_fit * x_line                    # 计算拟合线上对应的 y 值

plt.figure(figsize=(10, 6))                        # 新建画布
plt.scatter(x2, y2, color='#C49A6C', s=80, zorder=3, label='实际数据')   # 散点：zorder=3 画在图层上层
plt.plot(x_line, y_line, color='#8B6F4E', linewidth=2, label=f'拟合线: y={a_fit:.0f}+{b_fit:.2f}x')   # 拟合线
plt.xlabel('产量（件）')                            # x 轴
plt.ylabel('总成本（元）')                          # y 轴
plt.title('产量与总成本散点图')                      # 标题
plt.legend()                                      # 图例
plt.grid(True, alpha=0.3)                         # 网格
plt.show()                                        # 显示
print(f'拟合结果：a = {a_fit:.2f}（固定成本），b = {b_fit:.2f}（单位变动成本）')   # 打印经济含义
```

```python
# ===== 实战案例：广告费 vs 销售额 =====
ad = np.array([5, 8, 12, 15, 18, 22, 25, 30, 35, 40])       # 广告费（千元）
sales = np.array([50, 70, 90, 110, 125, 150, 165, 190, 210, 240])   # 销售额（千元）

coef2 = np.polyfit(ad, sales, 1)                  # 线性拟合
b_ad, a_ad = coef2[0], coef2[1]                   # 斜率与截距

ad_line = np.linspace(ad.min(), ad.max(), 100)    # 画线用的 x 序列
sales_line = a_ad + b_ad * ad_line                # 拟合线 y 值

plt.figure(figsize=(10, 6))                        # 画布
plt.scatter(ad, sales, color='#C49A6C', s=80, zorder=3, label='月度数据')   # 散点
plt.plot(ad_line, sales_line, color='#8B6F4E', linewidth=2, label='拟合线')   # 拟合线
plt.xlabel('广告费（千元）')                        # x 轴
plt.ylabel('销售额（千元）')                        # y 轴
plt.title('广告费与销售额关系')                      # 标题
plt.legend()                                      # 图例
plt.grid(True, alpha=0.3)                         # 网格
plt.show()                                        # 显示

pred_50 = a_ad + b_ad * 50                         # 预测广告费 50 千元时的销售额
print(f'投入 50 千元广告费 → 预测销售额 {pred_50:.1f} 千元')   # 打印预测结果
print('注意：50 超出数据范围（5~40），属于外推预测，风险更大')   # 外推风险提示
```

## 4. 本周小结

| 方法 | 公式 | 特点 |
|---|---|---|
| 两点式 | $b=\frac{y_2-y_1}{x_2-x_1}$，$a=y_1-bx_1$ | 精确（数据完全线性时）|
| 高低点法 | 用最高/最低两点 | 简单快速，但易受异常值影响 |
| 散点图+拟合 | `np.polyfit(x, y, 1)` | 利用全部数据（下周严格推导）|
